```python
import os
from dotenv import load_dotenv
from copy import deepcopy
load_dotenv()
os.chdir(os.getenv('PARENT_DIR'))
import json
```

```python
from glob import glob
files = glob('results/sequential_gemini/**/inference_results.json', recursive=True)
# Sort files by their parent directory name (which is the model name)
files.sort(key=lambda x: x.split('/')[2])
```

```python
files
```

```python
concat_results = []
for file in files:
	with open(file, 'r') as f:
		data = json.load(f)
		concat_results.extend(data)
```
