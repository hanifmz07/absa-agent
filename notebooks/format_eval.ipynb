```python
import os
from dotenv import load_dotenv
from copy import deepcopy
load_dotenv()
os.chdir(os.getenv('PARENT_DIR'))
```

```python
from glob import glob
import json
import pandas as pd
```

```python
paths = glob('results/async/Qwen3-8B/hotel_reviews/indo/mvp_aos/exp1/max_retries_*/seed_*/*/agent_evaluation_results.json')
paths
```

```python
data = {
	'max_retries': [],
	'seed': [],
	'precision': [],
	'recall': [],
	'f1': []
}
for path in paths:
	max_retries = path.split('/')[7].split('_')[-1]
	seed = path.split('/')[8].split('_')[-1]
	with open(path, 'r') as f:
		results = json.load(f)
	data['max_retries'].append(max_retries)
	data['seed'].append(seed)
	data['precision'].append(results['metrics']['precision'])
	data['recall'].append(results['metrics']['recall'])
	data['f1'].append(results['metrics']['f1'])
```

```python
df = pd.DataFrame(data)
df
```

```python
os.makedirs('temp', exist_ok=True)
df.to_csv('temp/eval.csv', index=False)
```
