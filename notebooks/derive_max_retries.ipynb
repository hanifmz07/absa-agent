```python
# result_paths = glob('results/async/Qwen3-8B/hotel_reviews/*/mvp_aos/exp1/max_retries_10/*/*/inference_results.json')
result_paths = glob('results/sequential_gemini/gemini-3-flash-preview/hotel_reviews/*/mvp_aos/exp1/max_retries_10/*/*/inference_results.json')
# Filter indo from results_paths
# result_paths = [path for path in result_paths if 'indo' not in path]
result_paths
```

```python
max_retries_to_derive = [1,3,5]
for result_path in result_paths:
	with open(result_path) as f:
		data = json.load(f)
	for max_retries in max_retries_to_derive:
		new_data = []
		for instance in data:
			new_instance = deepcopy(instance)
			current_history_len = len(instance['history'])
			if current_history_len > max_retries:
				new_instance['history'] = instance['history'][:max_retries]
				new_instance['final_extraction'] = new_instance['history'][-1]['extraction']
				new_instance['attempts'] = max_retries
				new_instance['status'] = 'failed'
			new_data.append(new_instance)
		new_result_path = result_path.replace('max_retries_10', f'max_retries_{max_retries}')
		os.makedirs(os.path.dirname(new_result_path), exist_ok=True)
		with open(new_result_path, 'w') as f:
			json.dump(new_data, f, indent=4, ensure_ascii=False)
		print(f"Derived {new_result_path} with max_retries={max_retries}")
		# print(len(new_data))
	# break
```
